# Digital Filters for Computer-Savvy Musicians

A digital filter in audio processing is both very simple and very complicated. It's the implementation that is simple. There are two kinds of filters we deal with, called finite impulse response (FIR) filters, and infinite impulse response (IIR) filters.

We'll be using some math notation here. First of all, we'll represent an audio signal as $$x[n]$$ where $n$ is the sample index. If our sample rate is 48 kHz, then $x[48000]$ will be the audio sample at precisely one second. We'll represent the output of the filter as $$y[n]$$ so it follows that we could write a simple equation that represents our filter: $$y[n] = f(x[n])$$
A simple FIR filter looks like this: $$y[n] = b_0 x[n] + b_1 x[n-1]$$ This equation means that the output of our filter is simply a weighted average of the current sample plus the previous sample. If $b_0 = b_1 = 0.5$, then this is just an average ($y[n] = 0.5x[n] + 0.5x[n-1]$). We call this a *two-tap* FIR filter. We can make filters using a lot more audio samples, of course. The general form of a FIR filter is $$y[n] = \sum_{i=0}^{B} b_i x[n-i]$$ where $B+1$ is the number of taps. If you're a little DSP-savvy, you'll recognize this is the discrete **convolution** operation.

A simple IIR filter looks like this: $$y[n] = b_0 x[n] + b_1 x[n-1] + a_1 y[n-1]$$ This means that we are adding the current audio sample, the previous audio sample, and the *previous filter output sample* (multiplied by weighting coefficients, of course). So, it seems that an IIR filter has feedback in it. (FIR filters can have feedback too, but you probably won't run into those.) If we have too much feedback in our filter, it can blow up and give us really loud audio. So an important part of IIR filter design involves managing the feedback so it doesn't get out of control.

We can generalize the IIR filter equation to
$$ y[n] = \sum_{i=0}^{B} b_i x[n-i] - \sum_{j=1}^{A} a_i y[n-i]$$
You'll notice that $j$ starts at 1. This is because we don't have access to the current filter output sample--we're computing it with this formula.

Most filters in audio synthesis are IIR filters.
